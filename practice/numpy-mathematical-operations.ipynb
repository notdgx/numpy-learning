## CLIP 


````python
x = np.array([-5, 2, 7, 12, 20])
print(x)
np.clip(x, 0, 10)
````

```
-5 → 0    because it is below 0
 2 → 2    already inside
 7 → 7    already inside
12 → 10   because it is above 10
20 → 10   because it is above 10
```

````python
x = np.array([10, 20, 30])
lower = np.array([5, 15, 25])
upper = np.array([15, 25, 15])

np.clip(x, lower, upper)
````

````python
np.clip(x, 5, None)
````
