```python
x = a>35
```

```python
a[x]
```

```python
print(a == 30)
print(a != 30)
print(a < 30)
print(a <= 30)
print(a > 30)
print(a >= 30)
```

```python
print(a[a == 30])
print(a[a != 30])
print(a[a < 30])
print(a[a <= 30])
print(a[a > 30])
print(a[a >= 30])
```

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])


print(a,"\n")
print("dim : ",a.ndim)
print("size : ",a.size)
print("shape : ",a.shape)
print("dtype : ",a.dtype)
```

```python
a = np.arange(24).reshape(2, 3, 4)
```

```python
print(a)
```

```python
print(a[:,[1,2]])
print("------------")
print(a[:,[1,2],:])
print("------------")
print(a[:,:,[1,2]])
print("------------")
print(a[[1,0]])
```

```python
a[:,[1,2]] = 999
print(a)
print("------------")
a[:,[1,2],:] = 0
print(a)
print("------------")
a[:,:,[1,2]] = -1
print(a)
print("------------")
a[[1,0]] = 666
print(a)
print("------------")
```

# WHERE 

```python
a = np.array([10, 20, 30, 40, 50])

np.where(a > 25)
```

```python
x = np.where(a > 25)[0]
print(a[[x]])
```

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
b = x,y = np.where(a > 25)

print(b)
print(a[[x],[y]])
print(a[[x],[y]].shape)
```

```python
print(a)
```

```python
print(np.where(a>50,1000,9999))
print(a)
```

```python
np.where(a > 25, a * 10, a)
```

```python
marks = np.array([35, 72, 48, 91, 20])

np.where(marks >= 50, "Pass", "Fail")
```

```python
temperature = np.array([15, 25, 35, 40])

np.where(temperature > 30, "Hot", "Normal")
```
