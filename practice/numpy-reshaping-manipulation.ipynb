# COMBINING ARRAYS 

```python
a = np.array([[1, 2],
              [3, 4]])

print(a,"\n")
print("Shape : ",a.shape)
print("Size : ",a.size)
print("Dtype : ",a.dtype)
print("Ndim : ",a.ndim,"\n")


b = np.array([[5, 6],
              [7, 8]])

print(b,"\n")
print("Shape : ",b.shape)
print("Size : ",b.size)
print("Dtype : ",b.dtype)
print("Ndim : ",b.ndim)
```

```python
print(np.concatenate((a,b),axis = 0))
print(np.concatenate((a,b),axis = 0).shape)
```

```python
print(np.concatenate((a,b),axis = 1))
print(np.concatenate((a,b),axis = 1).shape)
```

```python
a = np.array([[[1, 2],
              [3, 4]],
              [[1, 2],
              [3, 4]]])

print(a,"\n")
print("Shape : ",a.shape)
print("Size : ",a.size)
print("Dtype : ",a.dtype)
print("Ndim : ",a.ndim,"\n")


b = np.array([[[5, 6],
              [7, 8]],
              [[5, 6],
              [7, 8]]])

print(b,"\n")
print("Shape : ",b.shape)
print("Size : ",b.size)
print("Dtype : ",b.dtype)
print("Ndim : ",b.ndim)
```

```python
print(np.concatenate((a,b),axis = 1))
print(np.concatenate((a,b),axis = 1).shape)
```

```python
a = np.array([[1, 2],
              [3, 4]])

print(a,"\n")
print("Shape : ",a.shape)
print("Size : ",a.size)
print("Dtype : ",a.dtype)
print("Ndim : ",a.ndim,"\n")


b = np.array([[5, 6],
              [7, 8]])

print(b,"\n")
print("Shape : ",b.shape)
print("Size : ",b.size)
print("Dtype : ",b.dtype)
print("Ndim : ",b.ndim)
```

```python
print(np.stack((a, b), axis=0))
print(np.stack((a, b), axis=0).shape)
```
