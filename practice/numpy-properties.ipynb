# VIEW vs COPY vs ASSIGNMENT 

```python
a = np.array([10, 20, 30])
b = a
print(a,b," >")
b[0] = 45
print(a,b)

print(a is b)
print(np.shares_memory(a,b))
```

```python
a = np.array([10, 20, 30,40,50 ])
b = a[0:3]
print(a,b," >")
b[0] = 45
print(a,b)

print(a is b)
print(np.shares_memory(a,b))
```

```python
a = np.array([10, 20, 30])
b = a.copy()
print(a,b," >")
b[0] = 45
print(a,b)

print(a is b)
print(np.shares_memory(a,b))
```
