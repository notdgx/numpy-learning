```python
names = np.array(["Alice", "Bob", "Charlie", "David"])
marks = np.array([70, 95, 60, 85])

marks_sorted = np.sort(marks)
print(marks_sorted,"\n")
print("Shape ->",marks_sorted.shape,"\n")

names_sorted_index = np.argsort(marks)

print(names_sorted_index,"\n")
print("Shape ->",names_sorted_index.shape)


print(names[names_sorted_index], "\n")
print(marks_sorted, "\n")
```

```python
a = np.array([30, 10, 20, 10,30,10,10 ,3,32,3])
print([x for x in range(10)])
print(a,"\n")
b = np.sort(a,stable=True)
c = np.argsort(a,stable=True)
print(b,"\n")
print("Shape ->",b.shape,"\n")

print(c,"\n")
print("Shape ->",c.shape)
```

# UNIQUE 

```python
a = np.array([4, 2, 4, 3, 2, 4])
np.unique(a)
```

```python
a = np.array([
    [
        [90, 70, 50],
        [60, 40, 15]
    ],
    [
        [90, 70, 50],
        [15, 50, 15]
    ]
])

print(a,"\n")
print("Shape ->",a.shape)
print("Dtype ->",a.dtype)
```

```python
np.unique(a)
```

```python
np.unique(a,axis=0)
```

```python
np.unique(a,axis=1)
```

```python
a = np.array([
    [
        [90, 70, 50],
        [90, 70, 50]
    ],
    [
        [90, 70, 50],
        [15, 50, 15]
    ]
])
np.unique(a,axis=1)
```

```python
np.unique(a,axis=2)
```

```python
type(np.unique(a,return_index=True))
```

```python
np.unique(a,return_index=True)
```

```python
np.unique(a,return_index=True,return_inverse=True)
```

```python
np.unique(a,return_index=True,return_inverse=True,return_counts=True)
```

```python
b = [x for x in np.unique(a,return_index=True,return_inverse=True,return_counts=True)]
for i in b:
    print(i)
    print("Shape ->",i.shape,"\n")
```

# SEARCH SORTED

```python
a = np.array([10, 20, 30, 40])
```

```python
np.searchsorted(a, 25)
```

```python
a = np.array([40, 10, 30, 20])
np.searchsorted(a, 25)
```

```python
a = np.array([10, 20, 20, 20, 30])
np.searchsorted(a,20)
```

```python
np.searchsorted(a,20,side='right')
```

```python
left = np.searchsorted(a, 20, side='left')
right = np.searchsorted(a, 20, side='right')

print("Range : left-> ",left,"right ->",right)
```

```python
np.searchsorted(a, [5, 15, 35, 50],side="right")
```

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

np.searchsorted(a[0], 25)
```
